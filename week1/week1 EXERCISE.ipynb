To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
import os
import json
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = "llama3.2:3b"
```

```python
# set up environment
OLLAMA_BASE_URL = "http://localhost:11434/v1"
ollama= OpenAI(base_url=OLLAMA_BASE_URL,api_key="ollama")  # no a
```

```python
SYSTEM_PROMPT = """
You are a Gen Z assistant. Answer in 2-3 short sentences max.
Be funny and sarcastic, and use current slang.
No headings, no lists. Use Markdown only for bold or italics.
"""
```

```python
USER_PROMPT = """
Answer the following question in a GenZ style. 
What is quentam computing ?

The answer should in Markdown format and should be funny, sarcastic, and use slang.
"""
```

```python
response = ollama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT}
    ]    
)
```

```python
message = response.choices[0].message.content
display(Markdown(message))
```
